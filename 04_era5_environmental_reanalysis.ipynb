```python
REF_LAT = -69.4069
REF_LON = 76.1956

grid_rows = []

for lat in ds_wave.latitude.values:
    for lon in ds_wave.longitude.values:

        distance = (
            (float(lat) - REF_LAT)**2 +
            (float(lon) - REF_LON)**2
        ) ** 0.5

        hs = ds_wave.swh.sel(
            latitude=lat,
            longitude=lon
        )

        valid_pct = float(
            hs.notnull().mean() * 100
        )

        grid_rows.append({
            "Latitude": float(lat),
            "Longitude": float(lon),
            "Valid_%": valid_pct,
            "Distance_deg": distance,
            "Mean_Hs_m": float(hs.mean(skipna=True)),
            "Max_Hs_m": float(hs.max(skipna=True))
        })

era_grid = pd.DataFrame(grid_rows)

era_grid = era_grid.sort_values(
    ["Distance_deg"]
)

display(era_grid)
```

```python
import cdsapi
from pathlib import Path
import calendar

client = cdsapi.Client()

OUT_DIR = Path("ERA5_2025_PrydzBay")
OUT_DIR.mkdir(exist_ok=True)

LAT = -68.5
LON = 76.0
YEAR = 2025

variables = [
    "significant_height_of_combined_wind_waves_and_swell",
    "mean_wave_direction",
    "mean_wave_period",
    "2m_temperature",
    "surface_solar_radiation_downwards",
    "sea_ice_cover"
]

times = [f"{h:02d}:00" for h in range(24)]

for month in range(1, 13):

    month_str = f"{month:02d}"
    days_in_month = calendar.monthrange(YEAR, month)[1]

    days = [
        f"{d:02d}"
        for d in range(1, days_in_month + 1)
    ]

    target = OUT_DIR / f"ERA5_{YEAR}_{month_str}.zip"

    if target.exists() and target.stat().st_size > 1000:
        print(f"Skipping {month_str}: already downloaded")
        continue

    print(f"\nDownloading {YEAR}-{month_str}...")
    print(f"Days: {days_in_month}")

    client.retrieve(
        "reanalysis-era5-single-levels",
        {
            "product_type": "reanalysis",

            "variable": variables,

            "year": str(YEAR),
            "month": month_str,
            "day": days,
            "time": times,

            # Single ERA5 grid cell
            "area": [
                LAT, LON,
                LAT, LON
            ],

            "format": "netcdf"
        },
        str(target)
    )

    print(
        f"Completed {month_str} | "
        f"{target.stat().st_size / 1024:.1f} KB"
    )

print("\n🔥 FULL 2025 ERA5 DOWNLOAD COMPLETE.")
```

```python
import pandas as pd
import numpy as np
import xarray as xr

# Open datasets
wave = xr.open_dataset(
    extract_dir / "data_stream-wave_stepType-instant.nc",
    engine="h5netcdf"
)

oper = xr.open_dataset(
    extract_dir / "data_stream-oper_stepType-instant.nc",
    engine="h5netcdf"
)

accum = xr.open_dataset(
    extract_dir / "data_stream-oper_stepType-accum.nc",
    engine="h5netcdf"
)

# Convert each dataset to DataFrame
w = wave[["swh", "mwp", "mwd"]].squeeze().to_dataframe().reset_index()
o = oper[["t2m", "siconc"]].squeeze().to_dataframe().reset_index()
s = accum[["ssrd"]].squeeze().to_dataframe().reset_index()

# Keep only required columns
w = w[["valid_time", "swh", "mwp", "mwd"]]
o = o[["valid_time", "t2m", "siconc"]]
s = s[["valid_time", "ssrd"]]

# Merge
df = w.merge(o, on="valid_time").merge(s, on="valid_time")

# Rename
df = df.rename(columns={
    "valid_time": "Timestamp",
    "swh": "Hs_m",
    "mwp": "Tp_s",
    "mwd": "Wave_direction_deg",
    "t2m": "Temperature_K",
    "siconc": "Sea_ice_fraction",
    "ssrd": "SSRD_J_m2"
})

# Temperature conversion
df["Temperature_C"] = df["Temperature_K"] - 273.15

# Solar conversion for hourly engineering use
df["Solar_irradiance_Wm2"] = df["SSRD_J_m2"] / 3600.0

# Reorder
df = df[
    [
        "Timestamp",
        "Hs_m",
        "Tp_s",
        "Wave_direction_deg",
        "Temperature_C",
        "Sea_ice_fraction",
        "SSRD_J_m2",
        "Solar_irradiance_Wm2"
    ]
]

# Sort
df = df.sort_values("Timestamp").reset_index(drop=True)

print(df.head())
print("\nShape:", df.shape)
print("\nMissing values:")
print(df.isna().sum())

print("\nSummary:")
print(df.describe())
```

```python
import cdsapi
import zipfile
from pathlib import Path
import xarray as xr
import pandas as pd
import numpy as np
import time
import os

# ============================================================
# CONFIGURATION
# ============================================================

YEAR = 2025

BASE_DIR = Path("ERA5_2025_FULL")
RAW_DIR = BASE_DIR / "raw"
EXTRACT_DIR = BASE_DIR / "extracted"
CSV_DIR = BASE_DIR / "engineering"

RAW_DIR.mkdir(parents=True, exist_ok=True)
EXTRACT_DIR.mkdir(parents=True, exist_ok=True)
CSV_DIR.mkdir(parents=True, exist_ok=True)

client = cdsapi.Client()

# ============================================================
# MONTHS
# ============================================================

months = {
    1: 31,
    2: 28,
    3: 31,
    4: 30,
    5: 31,
    6: 30,
    7: 31,
    8: 31,
    9: 30,
    10: 31,
    11: 30,
    12: 31
}

# ============================================================
# DOWNLOAD + PROCESS EACH MONTH
# ============================================================

for month, ndays in months.items():

    mm = f"{month:02d}"

    zip_file = RAW_DIR / f"ERA5_{YEAR}_{mm}.zip"
    extract_dir = EXTRACT_DIR / f"{YEAR}_{mm}"
    csv_file = CSV_DIR / f"ERA5_{YEAR}_{mm}_engineering.csv"

    print("\n" + "="*70)
    print(f"MONTH: {YEAR}-{mm}")
    print("="*70)

    # --------------------------------------------------------
    # Skip if engineering CSV already exists
    # --------------------------------------------------------

    if csv_file.exists():

        print("Engineering CSV already exists:")
        print(csv_file)
        print("Skipping month.")
        continue

    # --------------------------------------------------------
    # DOWNLOAD
    # --------------------------------------------------------

    if not zip_file.exists():

        print("Downloading ERA5...")

        request = {
            "product_type": ["reanalysis"],
            "variable": [
                "significant_height_of_combined_wind_waves_and_swell",
                "mean_wave_direction",
                "mean_wave_period",
                "2m_temperature",
                "surface_solar_radiation_downwards",
                "sea_ice_cover"
            ],
            "year": [str(YEAR)],
            "month": [mm],
            "day": [f"{d:02d}" for d in range(1, ndays + 1)],
            "time": [f"{h:02d}:00" for h in range(24)],
            "area": [-68.5, 76.0, -68.5, 76.0],
            "data_format": "netcdf",
            "download_format": "unarchived"
        }

        try:

            client.retrieve(
                "reanalysis-era5-single-levels",
                request,
                str(zip_file)
            )

            print("Download complete.")

        except Exception as e:

            print("DOWNLOAD FAILED:")
            print(e)

            print("\nYou can simply rerun the cell.")
            break

    else:

        print("Raw ZIP already exists.")
        print("Skipping download.")

    # --------------------------------------------------------
    # EXTRACT
    # --------------------------------------------------------

    extract_dir.mkdir(parents=True, exist_ok=True)

    nc_files = list(extract_dir.glob("*.nc"))

    if len(nc_files) < 3:

        print("Extracting archive...")

        with zipfile.ZipFile(zip_file, "r") as z:

            print("Files:")
            for name in z.namelist():
                print(" ", name)

            z.extractall(extract_dir)

    else:

        print("NetCDF files already extracted.")

    # --------------------------------------------------------
    # IDENTIFY STREAMS
    # --------------------------------------------------------

    wave_file = extract_dir / "data_stream-wave_stepType-instant.nc"
    oper_file = extract_dir / "data_stream-oper_stepType-instant.nc"
    accum_file = extract_dir / "data_stream-oper_stepType-accum.nc"

    # --------------------------------------------------------
    # OPEN
    # --------------------------------------------------------

    print("Opening NetCDF streams...")

    wave = xr.open_dataset(
        wave_file,
        engine="h5netcdf"
    )

    oper = xr.open_dataset(
        oper_file,
        engine="h5netcdf"
    )

    accum = xr.open_dataset(
        accum_file,
        engine="h5netcdf"
    )

    # --------------------------------------------------------
    # DATAFRAMES
    # --------------------------------------------------------

    w = (
        wave[["swh", "mwp", "mwd"]]
        .squeeze()
        .to_dataframe()
        .reset_index()
    )

    o = (
        oper[["t2m", "siconc"]]
        .squeeze()
        .to_dataframe()
        .reset_index()
    )

    s = (
        accum[["ssrd"]]
        .squeeze()
        .to_dataframe()
        .reset_index()
    )

    # Keep only required fields

    w = w[
        ["valid_time", "swh", "mwp", "mwd"]
    ]

    o = o[
        ["valid_time", "t2m", "siconc"]
    ]

    s = s[
        ["valid_time", "ssrd"]
    ]

    # --------------------------------------------------------
    # MERGE
    # --------------------------------------------------------

    df = (
        w
        .merge(o, on="valid_time")
        .merge(s, on="valid_time")
    )

    # --------------------------------------------------------
    # RENAME
    # --------------------------------------------------------

    df = df.rename(columns={
        "valid_time": "Timestamp",
        "swh": "Hs_m",
        "mwp": "Tp_s",
        "mwd": "Wave_direction_deg",
        "t2m": "Temperature_K",
        "siconc": "Sea_ice_fraction",
        "ssrd": "SSRD_J_m2"
    })

    # --------------------------------------------------------
    # CONVERSIONS
    # --------------------------------------------------------

    df["Temperature_C"] = (
        df["Temperature_K"] - 273.15
    )

    df["Solar_irradiance_Wm2"] = (
        df["SSRD_J_m2"] / 3600.0
    )

    # --------------------------------------------------------
    # FINAL COLUMN ORDER
    # --------------------------------------------------------

    df = df[
        [
            "Timestamp",
            "Hs_m",
            "Tp_s",
            "Wave_direction_deg",
            "Temperature_C",
            "Sea_ice_fraction",
            "SSRD_J_m2",
            "Solar_irradiance_Wm2"
        ]
    ]

    df = (
        df
        .sort_values("Timestamp")
        .reset_index(drop=True)
    )

    # --------------------------------------------------------
    # QC
    # --------------------------------------------------------

    expected = ndays * 24

    print("\nQC:")
    print("Records:", len(df))
    print("Expected:", expected)
    print("Missing:", int(df.isna().sum().sum()))

    if len(df) != expected:
        print("WARNING: unexpected record count.")

    if df.isna().sum().sum() > 0:
        print("WARNING: missing values detected.")

    # --------------------------------------------------------
    # SAVE
    # --------------------------------------------------------

    df.to_csv(csv_file, index=False)

    print("\nSaved:")
    print(csv_file)

    print("\nWave:")
    print(
        f"Hs = {df.Hs_m.min():.3f}"
        f" – {df.Hs_m.max():.3f} m"
    )

    print(
        f"Tp = {df.Tp_s.min():.2f}"
        f" – {df.Tp_s.max():.2f} s"
    )

    print("\nSolar:")
    print(
        f"Max = "
        f"{df.Solar_irradiance_Wm2.max():.1f} W/m²"
    )

    # Close datasets

    wave.close()
    oper.close()
    accum.close()

    print("\nMonth complete.")

    # Small delay before next CDS request

    time.sleep(3)

print("\n" + "="*70)
print("2025 DOWNLOAD PROCESS FINISHED")
print("="*70)
```

```python
from pathlib import Path
import xarray as xr
import numpy as np
import pandas as pd

base = Path("ERA5_2025_FULL/extracted")

months = range(1, 10)

rows = []

for month in months:

    mm = f"{month:02d}"
    folder = base / f"2025_{mm}"

    wave_file = folder / "data_stream-wave_stepType-instant.nc"
    oper_file = folder / "data_stream-oper_stepType-instant.nc"
    accum_file = folder / "data_stream-oper_stepType-accum.nc"

    wave = xr.open_dataset(wave_file, engine="h5netcdf")
    oper = xr.open_dataset(oper_file, engine="h5netcdf")
    accum = xr.open_dataset(accum_file, engine="h5netcdf")

    hs = wave["swh"].squeeze().values
    tp = wave["mwp"].squeeze().values
    mwd = wave["mwd"].squeeze().values

    temp = oper["t2m"].squeeze().values
    ice = oper["siconc"].squeeze().values
    solar = accum["ssrd"].squeeze().values

    rows.append({
        "Month": mm,
        "Hours": len(hs),

        "Hs_valid": np.isfinite(hs).sum(),
        "Hs_missing": np.isnan(hs).sum(),

        "Tp_valid": np.isfinite(tp).sum(),
        "Tp_missing": np.isnan(tp).sum(),

        "Direction_valid": np.isfinite(mwd).sum(),
        "Direction_missing": np.isnan(mwd).sum(),

        "Temp_valid": np.isfinite(temp).sum(),
        "Ice_valid": np.isfinite(ice).sum(),

        "Solar_valid": np.isfinite(solar).sum(),

        "Hs_mean": np.nanmean(hs),
        "Hs_max": np.nanmax(hs)
    })

    wave.close()
    oper.close()
    accum.close()

coverage = pd.DataFrame(rows)

print(coverage.to_string(index=False))
```

```python
import copernicusmarine
import pandas as pd
import numpy as np
from pathlib import Path
import xarray as xr
import os

# ============================================================
# COPERNICUS WAVE COVERAGE TEST
# ============================================================

dataset_id = "cmems_mod_glo_wav_my_0.2deg_PT3H-i"

# Reference region around Bharati / Prydz Bay
latitudes = [-67.8, -68.2, -68.6, -69.0, -69.4]
longitudes = [75.4, 75.8, 76.2, 76.6, 77.0]

# Four representative periods
periods = {
    "JAN": ("2025-01-01T00:00:00", "2025-01-31T21:00:00"),
    "APR": ("2025-04-01T00:00:00", "2025-04-30T21:00:00"),
    "JUL": ("2025-07-01T00:00:00", "2025-07-31T21:00:00"),
    "OCT": ("2025-10-01T00:00:00", "2025-10-31T21:00:00"),
}

test_dir = Path("COPERNICUS_WAVE_COVERAGE_TEST")
test_dir.mkdir(exist_ok=True)

results = []

for lat in latitudes:

    for lon in longitudes:

        print("\n" + "="*60)
        print(f"CELL: {lat} , {lon}")
        print("="*60)

        cell_results = {
            "Latitude": lat,
            "Longitude": lon
        }

        for period_name, (start, end) in periods.items():

            filename = (
                test_dir /
                f"wave_{lat}_{lon}_{period_name}.nc"
            )

            try:

                if not filename.exists():

                    print(
                        f"Downloading {period_name}..."
                    )

                    copernicusmarine.subset(
                        dataset_id=dataset_id,
                        variables=[
                            "VHM0",
                            "VTM02",
                            "VMDR"
                        ],
                        minimum_longitude=lon,
                        maximum_longitude=lon,
                        minimum_latitude=lat,
                        maximum_latitude=lat,
                        start_datetime=start,
                        end_datetime=end,
                        output_filename=str(filename)
                    )

                ds = xr.open_dataset(filename)

                hs = ds["VHM0"].values
                tp = ds["VTM02"].values
                direction = ds["VMDR"].values

                hs = np.asarray(hs).flatten()
                tp = np.asarray(tp).flatten()
                direction = np.asarray(direction).flatten()

                valid_hs = np.isfinite(hs)
                valid_tp = np.isfinite(tp)
                valid_dir = np.isfinite(direction)

                n = len(hs)

                coverage = (
                    np.sum(valid_hs) / n * 100
                    if n > 0 else 0
                )

                cell_results[f"{period_name}_coverage_pct"] = coverage
                cell_results[f"{period_name}_Hs_mean"] = (
                    np.nanmean(hs)
                    if np.any(valid_hs) else np.nan
                )
                cell_results[f"{period_name}_Hs_max"] = (
                    np.nanmax(hs)
                    if np.any(valid_hs) else np.nan
                )

                print(
                    f"{period_name}: "
                    f"{coverage:.1f}% valid | "
                    f"Hs mean = "
                    f"{cell_results[f'{period_name}_Hs_mean']:.3f} m"
                )

                ds.close()

            except Exception as e:

                print(
                    f"{period_name}: FAILED"
                )

                print(e)

                cell_results[
                    f"{period_name}_coverage_pct"
                ] = np.nan

        results.append(cell_results)

coverage_df = pd.DataFrame(results)

print("\n\n" + "="*80)
print("COPERNICUS WAVE COVERAGE SUMMARY")
print("="*80)

display(coverage_df)
```
