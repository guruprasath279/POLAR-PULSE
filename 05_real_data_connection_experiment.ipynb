```python
import copernicusmarine
import xarray as xr
import numpy as np
import pandas as pd

LAT_OCEAN = -68.8
LON_OCEAN = 76.2

WAVE_FILE = "PrydzBay_Copernicus_Waves_2025.nc"

copernicusmarine.subset(
    dataset_id="cmems_mod_glo_wav_my_0.2deg_PT3H-i",

    variables=[
        "VHM0",   # Significant wave height
        "VTPK",   # Peak wave period
        "VMDR"    # Mean wave direction
    ],

    minimum_longitude=LON_OCEAN,
    maximum_longitude=LON_OCEAN,
    minimum_latitude=LAT_OCEAN,
    maximum_latitude=LAT_OCEAN,

    start_datetime="2025-01-01T00:00:00",
    end_datetime="2025-12-31T21:00:00",

    output_filename=WAVE_FILE
)

print("Download complete:", WAVE_FILE)
```

```python
ds = xr.open_dataset(WAVE_FILE)

print("=== DATASET DIMENSIONS ===")
print(ds.dims)

print("\n=== VARIABLES ===")
print(list(ds.data_vars))

print("\n=== TIME RANGE ===")
print(ds.time.values[0])
print(ds.time.values[-1])

print("\n=== RECORD COUNT ===")
print(len(ds.time))

# Extract variables
Hs = ds["VHM0"].values.flatten()
Tp = ds["VTPK"].values.flatten()
Dir = ds["VMDR"].values.flatten()

print("\n=== Hs ===")
print("Min :", np.nanmin(Hs))
print("Max :", np.nanmax(Hs))
print("Mean:", np.nanmean(Hs))
print("NaN :", np.isnan(Hs).sum())

print("\n=== Tp ===")
print("Min :", np.nanmin(Tp))
print("Max :", np.nanmax(Tp))
print("Mean:", np.nanmean(Tp))
print("NaN :", np.isnan(Tp).sum())

print("\n=== Direction ===")
print("Min :", np.nanmin(Dir))
print("Max :", np.nanmax(Dir))
print("Mean:", np.nanmean(Dir))
print("NaN :", np.isnan(Dir).sum())

print("\n=== FIRST 10 RECORDS ===")

df_wave = pd.DataFrame({
    "Timestamp": pd.to_datetime(ds.time.values),
    "Hs_m": Hs,
    "Tp_s": Tp,
    "Wave_direction_deg": Dir
})

display(df_wave.head(10))
```

```python
# ============================================
# V8.2 — Wave Data Availability Diagnostics
# ============================================

df_wave = pd.DataFrame({
    "Timestamp": pd.to_datetime(ds.time.values),
    "Hs_m": Hs,
    "Tp_s": Tp,
    "Wave_direction_deg": Dir
})

# Valid only when all three wave parameters exist
df_wave["Valid_wave"] = (
    df_wave["Hs_m"].notna() &
    df_wave["Tp_s"].notna() &
    df_wave["Wave_direction_deg"].notna()
)

# Month-wise availability
df_wave["Month"] = df_wave["Timestamp"].dt.month

monthly = df_wave.groupby("Month").agg(
    Total_records=("Timestamp", "count"),
    Valid_records=("Valid_wave", "sum")
)

monthly["Valid_%"] = (
    monthly["Valid_records"] /
    monthly["Total_records"] * 100
)

print("=== MONTHLY WAVE DATA AVAILABILITY ===")
display(monthly)

print("\n=== FIRST VALID RECORD ===")
print(df_wave.loc[df_wave["Valid_wave"], "Timestamp"].iloc[0])

print("\n=== LAST VALID RECORD ===")
print(df_wave.loc[df_wave["Valid_wave"], "Timestamp"].iloc[-1])

# Continuous valid periods
valid = df_wave[df_wave["Valid_wave"]].copy()

valid["gap_hours"] = (
    valid["Timestamp"].diff()
    .dt.total_seconds() / 3600
)

print("\n=== VALID RECORD GAP STATISTICS ===")
print(valid["gap_hours"].describe())

# Longest continuous valid/invalid stretches
df_wave["state_change"] = (
    df_wave["Valid_wave"] != df_wave["Valid_wave"].shift()
)

df_wave["block"] = df_wave["state_change"].cumsum()

blocks = df_wave.groupby("block").agg(
    Start=("Timestamp", "first"),
    End=("Timestamp", "last"),
    Valid=("Valid_wave", "first"),
    Records=("Timestamp", "count")
)

blocks["Duration_hours"] = blocks["Records"] * 3

print("\n=== VALID/INVALID BLOCKS ===")
display(blocks.head(30))
```

```python
print("\n=== VALID Hs BY MONTH ===")

monthly_hs = df_wave[df_wave["Hs_m"].notna()].groupby("Month")["Hs_m"].agg(
    Count="count",
    Min="min",
    Mean="mean",
    Max="max"
)

display(monthly_hs)
```

```python
import copernicusmarine
import xarray as xr
import numpy as np
import pandas as pd

# ============================================
# V8.2 — Regional Ocean Grid Availability Scan
# ============================================

TEST_FILE = "PrydzBay_wave_grid_scan_2025.nc"

copernicusmarine.subset(
    dataset_id="cmems_mod_glo_wav_my_0.2deg_PT3H-i",

    variables=["VHM0"],

    minimum_longitude=74.0,
    maximum_longitude=79.0,

    minimum_latitude=-70.5,
    maximum_latitude=-67.0,

    start_datetime="2025-01-01T00:00:00",
    end_datetime="2025-03-31T21:00:00",

    output_filename=TEST_FILE
)

scan_ds = xr.open_dataset(TEST_FILE)

print(scan_ds)
```

```python
# ============================================
# Calculate valid-data percentage per grid cell
# ============================================

hs_grid = scan_ds["VHM0"]

availability = (
    hs_grid.notnull()
    .mean(dim="time") * 100
)

mean_hs = hs_grid.mean(dim="time", skipna=True)

results = []

for lat in scan_ds.latitude.values:
    for lon in scan_ds.longitude.values:

        avail = float(
            availability.sel(
                latitude=lat,
                longitude=lon
            )
        )

        mean_value = float(
            mean_hs.sel(
                latitude=lat,
                longitude=lon
            )
        )

        results.append({
            "Latitude": float(lat),
            "Longitude": float(lon),
            "Valid_%": avail,
            "Mean_Hs_m": mean_value
        })

grid_df = pd.DataFrame(results)

# Distance from Bharati reference
REF_LAT = -69.4069
REF_LON = 76.1956

grid_df["Distance_deg"] = np.sqrt(
    (grid_df["Latitude"] - REF_LAT)**2 +
    (grid_df["Longitude"] - REF_LON)**2
)

grid_df = grid_df.sort_values(
    ["Valid_%", "Distance_deg"],
    ascending=[False, True]
)

print("=== BEST AVAILABLE OCEAN GRID CELLS ===")
display(grid_df.head(20))
```

```python
import copernicusmarine
import xarray as xr
import numpy as np
import pandas as pd

# ============================================
# V8.2 — Annual Candidate Cell Test
# ============================================

CANDIDATES = [
    (-67.6, 75.2),
    (-67.4, 75.0),
    (-67.4, 75.2),
    (-67.2, 75.0),
    (-67.0, 74.8),
]

TEST_FILE = "PrydzBay_candidate_cells_2025.nc"

copernicusmarine.subset(
    dataset_id="cmems_mod_glo_wav_my_0.2deg_PT3H-i",

    variables=["VHM0"],

    minimum_longitude=74.6,
    maximum_longitude=75.4,

    minimum_latitude=-67.8,
    maximum_latitude=-66.8,

    start_datetime="2025-01-01T00:00:00",
    end_datetime="2025-12-31T21:00:00",

    output_filename=TEST_FILE
)

ds_candidates = xr.open_dataset(TEST_FILE)

print(ds_candidates)
```

```python
# ============================================
# Annual availability of candidate cells
# ============================================

rows = []

for lat, lon in CANDIDATES:

    series = ds_candidates["VHM0"].sel(
        latitude=lat,
        longitude=lon,
        method="nearest"
    )

    valid = series.notnull()

    rows.append({
        "Requested_lat": lat,
        "Requested_lon": lon,
        "Actual_lat": float(series.latitude.values),
        "Actual_lon": float(series.longitude.values),
        "Valid_records": int(valid.sum()),
        "Total_records": len(series.time),
        "Valid_%": float(valid.mean() * 100),
        "Mean_Hs_m": float(series.mean(skipna=True)),
        "Max_Hs_m": float(series.max(skipna=True))
    })

candidate_df = pd.DataFrame(rows)

print("=== 2025 ANNUAL WAVE AVAILABILITY ===")
display(candidate_df)
```
